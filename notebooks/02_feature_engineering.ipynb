```python
import pandas as pd

df = pd.read_csv('../data/processed/heart_disease_cleaned.csv')
df.head()
```

```python
# Composite Lifestyle Risk Score
df['lifestyle_risk_score'] = (
    df['Smoker'] +
    (1 - df['PhysActivity']) +
    (1 - df['Fruits']) +
    (1 - df['Veggies']) +
    (df['HvyAlcoholConsump']) 

)

df['lifestyle_risk_score'].value_counts()

```

```python
# Comorbidity Count

df['has_diabetes'] = (df['Diabetes'] > 0).astype(int)
df['comorbidity_count'] = df['HighChol'] + df['has_diabetes']

df['comorbidity_count'].value_counts()
```

```python
# PH-Aligned BMI Category

def bmi_category_ph(bmi):
    if bmi < 18.5:
        return 0
    elif bmi < 23:
        return 1
    elif bmi < 27:
        return 2
    else:
        return 3

df['bmi_cat_ph'] = df['BMI'].apply(bmi_category_ph)
```
